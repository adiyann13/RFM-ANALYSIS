```python
ecom_Data_new = ecom_Data_new.groupby("CustomerID").agg({'RECENCY':lambda x:(new_date-x.max()).days,
                                                        'MONETORY':lambda x : x.sum(),
                                                        'FREQUENCY': lambda x : x.count()
    
})
```

```python
## based on the core lets do a quantile cut
quantile = ecom_Data_new.quantile(q=[0.25,0.50,0.75])
quantile.to_dict()
```

```python
### now giving a score for recency frequency and monetory
# lower the recency better the score 
# higher the frequency better the score
```

```python
def recency_Score(x,q,d):
    if x<=d[q][0.25]:
        return 1
    elif x<=d[q][0.50]:
        return 2
    elif x<=d[q][0.75]:
        return 3
    else:
        return 4
    
    
```

```python
def freq_monet_score(x,q,d):
    if x<=d[q][0.25]:
        return 4
    elif x<=d[q][0.50]:
        return 3
    elif x<=d[q][0.75]:
        return 2
    else:
        return 1
```

```python
ecom_Data_new["Recency_score"] = ecom_Data_new["RECENCY"].apply(recency_Score,args=("RECENCY",quantile))
ecom_Data_new["frequency_Score"] = ecom_Data_new["FREQUENCY"].apply(freq_monet_score,args=("FREQUENCY",quantile))
ecom_Data_new["monetory_Score"] = ecom_Data_new["MONETORY"].apply(freq_monet_score,args=("MONETORY",quantile))
```

```python
ecom_Data_new
```

```python
## ading the recency frequency monetory score al togehter
ecom_Data_new["RFM_SCORE"] = ecom_Data_new[["Recency_score","frequency_Score","monetory_Score"]].sum(axis=1)
```

```python
ecom_Data_new
```

```python
## based on this score giving a loyality badge
loyality_badges = ["PLATINUM","GOLD","SILVER","BRONZE"]
ecom_Data_new["Loyality_Badge"] = pd.qcut(ecom_Data_new["RFM_SCORE"],q=4,labels =loyality_badges)
```

```python
ecom_Data_new
```

```python
# dropping columns which arent required
ecom_Data_new.drop(columns=["Recency_score","frequency_Score","monetory_Score","RFM_SCORE"],axis=1,inplace=True)
```
